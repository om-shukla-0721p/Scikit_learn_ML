```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.cluster import k_means
```

```python
class k_means_own:
    def __init__(self,n=2,max_iteration=100):
        self.n_cluster=n
        self.max_iteration=max_iteration
        self.centroids=None

    def assign_clusters(self,X):
        cluster_group=[]

        for row in X:
            distances=[]
            for centroid in self.centroids:
                vectors=row-centroid # calcauing (x1-x2),(y2-y1),(z2,z1)....
                # auto euclidian distance calculator 
                # with the help of manipulation to vectors
                distances.append((np.sqrt(np.dot(vectors,vectors))))
            min_dist=min(distances)
            min_index=distances.index(min_dist)
            cluster_group.append(min_index)

        return np.array(cluster_group)

    def move_centroids(self,X,cluster_group):
        new_centroids=[]

        # cluster_type=np.unique(cluster_group)

        for type in range(self.n_cluster):
            points=X[cluster_group==type]

            if len(points)==0: # handles if a centroid has no element in it 
                new_centroids.append(self.centroids[type])
            else:
                new_centroids.append(points.mean(axis=0))

        return np.array(new_centroids)


    def fit_predict(self,X):
        random_centroids=np.random.choice(0,X.shape[0],self.n_cluster,replace=False)
        self.centroids=X[random_centroids]

        for i in range(self.max_iteration):
            # assign cluster
            cluster_group=self.assign_clusters(X)

            old_centroids=self.centroids
            # move centroids
            new_centroids=self.move_centroids(X,cluster_group)
            self.centroids=new_centroids

            # check finish
            if (old_centroids==new_centroids).all():
                break

        return cluster_group

    
```
