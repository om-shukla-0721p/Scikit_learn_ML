```python
import numpy as np
import pandas as pd
```

```python
class own_logistic_reg:
    def __init__(self,learning_rate=0.01,epochs=100):
        self.coef=None
        self.lr=learning_rate
        self.epoch=epochs

    def sigmoid_func(self,val):
        return (1)/(1+np.e**(-val))

    def fit(self,X,y):
        # initializing coefs
        self.coef=np.ones(X.shape[1]+1)

        # insering ones col in coefs to handle constant(bias) easily
        X=np.insert(X,0,1,axis=1) # arr,col,val,axis

        for i in range(self.epoch):
            ind=np.random.randint(0,X.shape[0])

            y_pred=self.sigmoid_func((np.dot(X[ind],self.coef)))

            self.coef=self.coef+(self.lr*(y[ind]-y_pred)*X[ind])

            print(self.coef[0],self.coef[1:])


    def predict(self,X):
        # if u want ans in terms of probability
        # X=np.insert(X,0,1,axis=1)
        # return self.sigmoid_func(np.dot(X,self.coef))
        X=np.insert(X,0,1,axis=1)
        return np.dot(X,self.coef)
    
```
