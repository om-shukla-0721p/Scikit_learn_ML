```python
import numpy as np
import pandas as pd
```

```python
class Logistic_reg_GD_own:
    def __init__(self,learning_rate=0.01,epochs=100):
        self.coef=None
        self.lr=learning_rate
        self.epoch=epochs
        # intercept is handled with coefs only as self.coef[0]
    
    def sigmoid_func(self,val):
        return (1/(1+np.exp(-val)))

    def fit(self,X,y):
        # initializing coefs
        # +1 becuase taking self.coef[0] as bias
        self.coef=np.ones(X.shape[1]+1)
        

        # inserting 1's in X mat to make hadling intercept easier
        X=np.insert(X,0,1,axis=1)

        # Batch Gd
        for i in range(self.epoch):
            # prerequisite
            y_pred=self.sigmoid_func(np.dot(X,self.coef))

            # updaitng intercept and self.coef
            coef_der=(-1)*np.dot(X.T,y-y_pred)
            self.coef=self.coef-(self.lr*coef_der)
            # quicker way to write
            # self.coef=self.coef-(self.lr*((-1)*np.dot(X.T,y-y_pred))

    def predict(self,X):
        # returns probab of success
        X=np.insert(X,0,1,axis=1)
        return self.sigmoid_func(np.dot(X,self.coef))

        # to return True and False as ans
        # return self.sigmoid_func(np.dot(X,self.coef))>=0.5

```
