```python
import numpy as np
import pandas as pd
import random
```

### batch GD -> updates 1 time each epoch after seeing every row in the dataset
### SGD-> updates n(rows) times each epoch after seeing one row at once each time
### mini batch GD-> updates a specific number of times in every epoch after seeing a specific number of rows each time in the dataset

```python
class Mini_batch_GD_own:
    def __init__(self,batch_size=1,learning_rate=0.01,epochs=100):
        self.intercept=0
        self.coef=None
        self.batch_size=batch_size
        self.lr=learning_rate
        self.epoch=epochs
        

    def fit(self,X,y):
        if self.batch_size>X.shape[0]:
            self.batch_size=X.shape[0]
        # initializing intercept and coefs
        self.coef=np.ones(X.shape[1])
        self.intercept=0

        for i in range(self.epoch):
            for j in range(X.shape[0]//self.batch_size):
                ind=random.sample(range(0,X.shape[0]),self.batch_size)

                # prerequisites
                y_pred=np.dot(X[ind],self.coef)+self.intercept


                # updating intercept
                intercept_der=(-2/self.batch_size)*np.sum(y[ind]-y_pred)
                self.intercept=self.intercept-(self.lr*(intercept_der))

                # updating coefs
                slope_der=(-2/self.batch_size)*np.dot(X[ind].T,y[ind]-y_pred)
                self.coef=self.coef-(self.lr*slope_der)

                print(self.intercept,self.coef)
    def predict(self,X):
        return np.dot(self.coef.T,X)+self.intercept
                
```
