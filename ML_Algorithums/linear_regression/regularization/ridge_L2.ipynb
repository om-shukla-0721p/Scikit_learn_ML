```python
import numpy as np
import pandas as pd
```

```python
class ridge_reg_own:
    def __init__(self,learnning_rate=0.01,epochs=100,alpha=1):
        self.coef=None
        self.intercept=0
        self.epoch=epochs
        self.lr=learnning_rate
        self.alpha=alpha

    def fit(self,X,y):
        self.coef=np.ones(X.shape[1])
        self.intercept=0

        for i in range(self.epoch):
            for j in range(X.shape[0]):
                ind=np.random.randint(0,X.shape[0])

                y_pred=np.dot(X[ind],self.coef)+self.intercept

                # updating intercept
                intercept_der=(-2/1)*(y[ind]-y_pred)
                self.intercept=self.intercept-(self.lr*intercept_der)

                # updating slope
                slope_der=(-2/1)*(y[ind]-y_pred)*(X[ind])+(2*self.alpha*self.coef)
                self.coef=self.coef-(self.lr*slope_der)
    def predict(self,X):
        return np.dot(X,self.coef)+self.intercept
```

```python
import numpy as np

X = np.array([
    [1],
    [2],
    [3],
    [4],
    [5],
    [6],
    [7],
    [8]
])

y = np.array([
    5,
    8,
    11,
    14,
    17,
    20,
    23,
    26
])

model = ridge_reg_own(
    learnning_rate=0.001,
    epochs=1000,
    alpha=0.1
)

model.fit(X, y)

print("Coefficient:", model.coef)
print("Intercept:", model.intercept)

print("Prediction for x=10:", model.predict(np.array([[10]])))
```
