```python
import numpy as np 
import pandas as pd 
```

```python
# not proper 
# can get a zero weight to move again
class lasso_reg_own:
    def __init__(self,learning_rate=0.01,epochs=100,alpha=1):
        self.intercept=0
        self.coef=None
        self.lr=learning_rate
        self.epoch=epochs
        self.alpha=alpha

    def fit(self,X,y):
        self.intercept=1
        self.coef=np.ones(X.shape[1])

        for i in range(self.epoch):
            # prerequisite
            y_pred=np.dot(X,self.coef)+self.intercept

            # updating intercept 
            intercept_der=(-2)*np.sum(y-y_pred)
            self.intercept=self.intercept-(self.lr*(intercept_der))

            # updating slope
            # slope_der can get weight moving even after it turns zero
            slope_der=(-2)*np.dot(X.T,y-y_pred)+(self.alpha*np.sign(self.coef))
            self.coef=self.coef-(self.lr*(slope_der))

    def predict(self,X):
        return np.dot(X,self.coef)+self.intercept
    
```

```python
# poper implementation
class proper_lasso_own:
    def __init__(self,learning_rate=0.01,epochs=100,alpha=1):
            self.intercept=0
            self.coef=None
            self.lr=learning_rate
            self.epoch=epochs
            self.alpha=alpha
    
    def fit(self,X,y):
        self.intercept=1
        self.coef=np.ones(X.shape[1])

        for i in range(self.epoch):

            y_pred = np.dot(X, self.coef) + self.intercept

            # intercept
            intercept_der = (-2) * np.sum(y - y_pred)
            self.intercept = self.intercept - self.lr * intercept_der

            # normal LR gradient
            slope_der = (-2) * np.dot(X.T, y - y_pred)
            self.coef = self.coef - self.lr * slope_der

            # Lasso proximal step
            self.coef = np.sign(self.coef) * np.maximum(
                np.abs(self.coef) - self.lr * self.alpha, 0
            )

    def predict(self,X):
        return np.dot(X,self.coef)+self.intercept
    
    
```
