2) EDA on numerical Data
<br>a) Histograms

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv(r"d:\Downloads_ML_aaa\train.csv")
df.head()
```

```python
plt.hist(df["Age"],color="lime",edgecolor="black")
plt.title("people of titanic",fontsize=30,color="cyan")
plt.xlabel("Age",fontsize=20,color="violet")
plt.ylabel("number of people",fontsize=20,color="purple")
```

b) displot

```python
sns.displot(df["Age"],kde=True)
```

c) Boxplot

```python
sns.boxplot(df["Age"])
```
