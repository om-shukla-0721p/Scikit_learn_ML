```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

tips_data=sns.load_dataset("tips")
iris_data=sns.load_dataset("iris")
flights_data=sns.load_dataset("flights")
titanic_data=pd.read_csv(r"d:\Downloads_ML_aaa\train.csv")
# titanic_data.head()
# tips_data.head()
# iris_data.head()
# flights_data.head()
```

<br>a) scatter plot for (Numerical to NUmerical datasets)

```python
# with only 2 things to comapre it bivariate with more is multivariate
sns.relplot(tips_data,x="total_bill",y="tip",
            hue="sex",size="size",style="smoker",
            col="day",col_wrap=2,kind="scatter")
```

2) Barplot(Numerical-Categorical)

```python
sns.catplot(titanic_data,x="Pclass",y="Fare",hue="Sex",kind="bar")
# titanic_data.head()
```

3) Boxplot (Numerical to Categorical)

```python
sns.catplot(titanic_data,x="Sex",y="Age",
            col="Embarked",col_wrap=2,kind="box",
            hue="Survived")
# sns.boxplot(x=titanic_data["Sex"],y=titanic_data["Age"])
```

4) Displot (Numerical to Categorical)

```python
sns.kdeplot(titanic_data[titanic_data["Survived"]==0]["Age"],label="dead")
sns.kdeplot(titanic_data[titanic_data["Survived"]==1]["Age"],label="alive")
plt.legend()
plt.show()
```
