```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 

from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import KBinsDiscretizer
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv(r"d:\Downloads_ML_aaa\train.csv",usecols=["Age","Fare","Survived"])
```

```python
df.dropna(inplace=True)
df.isnull().sum()
df.head()
```

```python
X=df.iloc[:,1:3]
y=df.iloc[:,0]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=23)
```

```python
clf1=DecisionTreeClassifier()
clf1.fit(X_train,y_train)
y_pred1=clf1.predict(X_test)
```

```python
accuracy_score(y_test,y_pred1)
```

```python
np.mean(cross_val_score(DecisionTreeClassifier(),X,y,cv=10,scoring="accuracy"))
```

```python
kbin_age=KBinsDiscretizer(n_bins=10,encode="ordinal",strategy="uniform")
kbin_fare=KBinsDiscretizer(n_bins=10,encode="ordinal",strategy="uniform")
```

```python
trf=ColumnTransformer([
    ("Age",kbin_age,[0]),
    ("Fare",kbin_age,[1])
    ]
)
```

```python
X_train_transformed=trf.fit_transform(X_train)
X_test_transformed=trf.transform(X_test)
```

```python
trf.named_transformers_["Age"].n_bins_
trf.named_transformers_["Fare"].bin_edges_
```

```python
clf2=DecisionTreeClassifier()
clf2.fit(X_train_transformed,y_train)
y_pred2=clf2.predict(X_test_transformed)

print("accuracy_score:",accuracy_score(y_pred2,y_test))
```
