standardization-> Also called as Z-Score normalization

<br>
-- understand this <br>
Z score-> it is the amount of times the standard_diviation -- a value is above or below mean


steps <br>
a) take your col<br>
b) calculate the mean and standard diviation<br>
c) calculate the z score<br>

standard diviation formula=sqaure root [summation from 0 to N [(xi-u)**2]/mean]
<br>
z score=(x-mean)/standad diviation

After standardization <br>
mean of new standardized col becomes 0
<br> and its standard diviation is 1
<br> Always preserves the shape of the distribution

standardization does mean centering and scales the range of data

<br> it doesnt handles outliers
