## Normilation-> its quite in the name
<br> it helps us eliminate the unit and
<br> get data in a commom scale


### Types of normilization (most common ones)
1) min max normalization
2) mean normalization
3) max absolute normzalization
4) robust scaling

### 1) min max normalization
<br>
Formula=(x-min)/(max-min)
<br> in words-> how much the value is wrt to total range
<br> Range of new data is from [0,1]
<br> Preserves the shape of the distribution 

## 2) mean normalization

formula =(x-mean)/(max-min)
<br> in words-> how far we are from the mean wrt total range
<br> mean centering is dont because of (x-mean)
<br> range of the values is (-1,1)
<br>

## 3) max absolute normalization

formula=(x)/(|max|)<br>
where (|max|)--> max absolute value in the dataset<br>
no centering <br>
scales data in the range-> [-1,1]


## 4) robust scaling
<br> specifically made to handle outliers 

fromula=(x-midian)/(Q3-Q1)
<br>in words-> how far is the point from median measured in units of typical
<br>           middle-50% spread 
<br> Q3->75th percentile , Q1-> 25th percentile
<br> [Q3-Q1]--> interquartile range
<br> range-> (-R,R)
