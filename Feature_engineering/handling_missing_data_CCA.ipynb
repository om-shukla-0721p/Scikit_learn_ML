```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"d:\Downloads_ML_aaa\data_science_job.csv")
df.info()
df.isnull().sum()
```

```python
df.head()
```

```python
plt.figure(figsize=(12,6))
sns.countplot(df,x="job_title")
plt.xticks(rotation=90)
# plt.tight_layout()
```

```python
plt.figure(figsize=(30,12))
g=sns.catplot(df,x="job_title",y="salary",col="experience_level",col_wrap=2,kind="bar")

for ax in g.axes.flat:
    plt.sca(ax)
    plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.catplot(df,x="salary",y="job_title",col="experience_level",col_wrap=2,kind="box")
```
