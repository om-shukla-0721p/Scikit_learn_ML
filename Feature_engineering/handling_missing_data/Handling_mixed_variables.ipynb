```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv(r"d:\Downloads_ML_aaa\train.csv",usecols=["Ticket","Cabin"])
```

```python
df["Cabin"].unique()
df["Cabin_numerical"]=df["Cabin"].str.extract("(\d+)") # extracts the numeric part
df["Cabin_categorical"]=df["Cabin"].str[0] # extracts the categorical part
df.head(10)
```

```python
import seaborn as sns
sns.catplot(df["Cabin_categorical"],kind="bar")
```

```python
df["Ticket_num"]=df["Ticket"].apply(lambda x: x.split()[-1]) # spliting data
# converting data into numeric form 
df["Ticket_num"]=pd.to_numeric(df["Ticket_num"],errors="coerce",downcast="integer")

# spliting categorical part
df["Ticket_cat"]=df["Ticket"].apply(lambda x: x.split()[0])
print(df["Ticket_cat"])
# finishing-> removing the numerical part which came
df["Ticket_cat"]=np.where(df["Ticket_cat"].str.isdigit(),np.nan,df["Ticket_cat"])

df.head()
```
