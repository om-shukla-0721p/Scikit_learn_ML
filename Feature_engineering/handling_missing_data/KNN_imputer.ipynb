```python
from sklearn.model_selection import train_test_split
from sklearn.impute import KNNImputer,SimpleImputer
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
```

```python
df=pd.read_csv(r"d:/Downloads_ML_aaa/train.csv",usecols=["Age","Pclass",
                                                         "Fare","Survived"])
```

```python
df.head()
X=df.iloc[:,1:]
y=df["Survived"]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=23)
```

```python
knn1=KNNImputer(n_neighbors=5,weights="distance")
X_train_trf=knn1.fit_transform(X_train)
X_test_trf=knn1.transform(X_test)
```

```python
lr=LogisticRegression()
lr.fit(X_train_trf,y_train)
y_predict=lr.predict(X_test_trf)

print("Accuracy_score_lr:",accuracy_score(y_test,y_predict))
```

```python
# comparing with si
si=SimpleImputer()
X_train_trf2=si.fit_transform(X_train)
X_test_trf2=si.transform(X_test)

lr.fit(X_train_trf2,y_train)
y_predict=lr.predict(X_test_trf2)


print("Accuracy_score:",accuracy_score(y_predict,y_test))
```
