## Working With Dates

```python
orders["date"]=pd.to_datetime(orders["date"]) # converitng col in datetime
```

```python
# extracting year
orders["date_year"]=orders["date"].dt.year
# extracting month
orders["date_month"]=orders["date"].dt.month
# for month name 
#orders["date_month"]=orders["date"].dt.month_name()
# extracting day
orders["date_day"]=orders["date"].dt.day

# day of week
# orders["Week_day_num"]=orders["date"].dt.day_of_week
orders["week_day"]=orders["date"].dt.day_name()

# weekend or not
orders["weekend"]=np.where(orders["week_day"].isin(["Sunday","Friday","Saturday"]),1,0)

# week number of the year
orders["week_number"]=orders["date"].dt.isocalendar().week

# quarter of year
orders["year_quarter"]=orders["date"].dt.quarter
```
