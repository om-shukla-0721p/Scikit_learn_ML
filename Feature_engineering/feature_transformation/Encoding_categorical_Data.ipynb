## ordinal encoding
<br> generally for categorical input cols 
<br> when we have some kind of order in the col 
<br> for eg: education=[school,bachelor,master,phD]
<br> here school could be 0 , bachelor 1 ,master 2 and phD 3


## label encoding 
<br> generally for [Categorical output]
<br> very similar to ordinal encdoing
<br> for eg--> colors=[cat,dog]
<br> their is no order but classes like cat and dogs
<br> here no order is present 
<br> so we just give some numerical value to represent them
<br> like cat=1,dog=2

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df=pd.read_csv(r"d:\Downloads_ML_aaa\WA_Fn-UseC_-Telco-Customer-Churn.csv")

df=df.drop(["customerID","MultipleLines","DeviceProtection"],axis=1)
df=df.iloc[:,-5:]
```

```python
sns.countplot(df["PaymentMethod"])
```

```python
sns.displot(df,x="MonthlyCharges",kde=True)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df.iloc[:,0:2],df.iloc[:,-1],test_size=0.2)
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe=OrdinalEncoder(categories=[["No","Yes"],
                  ["Mailed check","Electronic check","Bank transfer (automatic)","Credit card (automatic)"]])
```

```python
oe.fit(X_train)

X_train=oe.transform(X_train)
X_test=oe.transform(X_test)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
le.fit(y_train)
le.classes_
```

```python
y_train=le.transform(y_train)
y_test=le.transform(y_test)
```
