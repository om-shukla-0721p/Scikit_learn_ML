```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
```

```python
df=pd.read_csv(r"d:/Downloads_ML_aaa/train.csv",usecols=["Age","Parch","SibSp",
                                                         "Survived","Pclass"])
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
X=df.iloc[:,1:]
y=df["Survived"]
np.mean(cross_val_score(LogisticRegression(),X,y,scoring="accuracy",cv=20))
```

## Applying feature Construction

```python
X["Family_size"]=X["SibSp"]+X["Parch"]+1
```

```python
# making the function
def family_size(num):
    if num==1:
        # alone
        return 0
    elif num<=4:
        return 1
    else:
        return 2
    
```

```python
X["Family_type"]=X["Family_size"].apply(family_size)
```

```python
X.head()
```

```python
X.drop(columns=["SibSp","Parch","Family_size"])
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring="accuracy",cv=20))
```
